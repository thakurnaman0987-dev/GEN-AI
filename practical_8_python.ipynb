```python
import numpy as np

customers = pd.DataFrame({
    "Customer_ID": range(101, 111),
    "Name": ["Alice", "Bob", "Carol", "Dan", "Eve",
             "Frank", "Grace", "Heidi", "Ivan", "Judy"],
    "Age": [25, 34, 29, 41, 38, 23, 45, 31, 28, np.nan],
    "Gender": ["F", "M", "F", "M", "F", "M", "F", "F", "M", "F"],
    "Purchase_Amount": [250.0, 400.0, 150.0, 300.0, np.nan,
                        120.0, 500.0, 220.0, 180.0, 350.0]
})

print(customers.head())
print(customers.describe())

# Check missing values
print(customers.isnull().sum())

# Handle missing values: fill with column mean
customers["Age"] = customers["Age"].fillna(customers["Age"].mean())
customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(
    customers["Purchase_Amount"].mean())
print(customers.isnull().sum())  # all 0 now
```

```python
print(customers[customers["Age"] > 30])
print(customers.groupby("Gender")["Purchase_Amount"].sum())
print(customers[["Customer_ID", "Name", "Purchase_Amount"]])
```

```python
customers["Discounted_Price"] = customers["Purchase_Amount"] * 0.9
print(customers.groupby("Gender")["Purchase_Amount"].mean())
sorted_df = customers.sort_values("Purchase_Amount", ascending=False)
print(sorted_df)
```
